```python
from pathlib import Path
from PIL import Image, ImageDraw, ImageFont

# 图片所在文件夹
results_dir = Path(__file__).resolve().parent / "results"

# 四张图片，按 2×2 顺序排列
files = [
    "figure_2a.png",
    "figure_2b.png",
    "figure_4a.png",
    "figure_4b.png",
]

labels = ["(a) Figure 2(a)", "(b) Figure 2(b)",
          "(c) Figure 4(a)", "(d) Figure 4(b)"]

# 读取图片
images = [
    Image.open(results_dir / name).convert("RGB")
    for name in files
]

# 统一子图尺寸，保留原始分辨率
cell_w = max(img.width for img in images)
cell_h = max(img.height for img in images)

margin = 40
title_h = 55

# 创建白色背景的大图
canvas_w = 2 * cell_w + 3 * margin
canvas_h = 2 * (cell_h + title_h) + 3 * margin

canvas = Image.new("RGB", (canvas_w, canvas_h), "white")
draw = ImageDraw.Draw(canvas)

try:
    font = ImageFont.truetype("DejaVuSans.ttf", 36)
except OSError:
    font = ImageFont.load_default()

# 将四张图片依次放到 2×2 网格
for i, (img, label) in enumerate(zip(images, labels)):
    row = i // 2
    col = i % 2

    x0 = margin + col * (cell_w + margin)
    y0 = margin + row * (cell_h + title_h + margin)

    # 子图标签
    draw.text((x0, y0), label, fill="black", font=font)

    # 居中放置图片
    x = x0 + (cell_w - img.width) // 2
    y = y0 + title_h + (cell_h - img.height) // 2

    canvas.paste(img, (x, y))

# 保存结果
png_path = results_dir / "figure_2_4_combined.png"
pdf_path = results_dir / "figure_2_4_combined.pdf"

canvas.save(png_path, dpi=(300, 300))
canvas.save(pdf_path, resolution=300.0)

print(f"PNG saved: {png_path}")
print(f"PDF saved: {pdf_path}")
```
